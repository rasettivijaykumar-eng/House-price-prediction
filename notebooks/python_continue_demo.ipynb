# Python continue statement demo

This notebook explains and demonstrates the `continue` statement in Python loops.

## 1. Create a Simple For Loop with continue

Write a basic for loop that prints numbers from 1 to 10, but skips the value 5 using the `continue` statement. Explain how `continue` causes the loop to move to the next iteration without executing the remaining code in the current iteration.

```python
for number in range(1, 11):
    if number == 5:
        continue
    print(number)

print("\nExplanation:")
print("The continue statement skips the current iteration when number == 5.")
print("The loop immediately moves to the next number without executing the remaining code in that iteration.")
```

## 2. Skip Even Numbers in a Range

Use a for loop over a range of values and skip even numbers with `continue`. Show how this produces only odd numbers and explain the control flow difference between `continue` and a conditional `if` statement.

```python
for value in range(1, 11):
    if value % 2 == 0:
        continue
    print(value)

print("\nOnly odd numbers are printed because even values are skipped.")
print("Unlike a simple if block, continue does not run the rest of the loop body for that iteration.")
```

## 3. Use continue in a While Loop

Create a while loop that increments a counter and prints values except when the counter is a specific number. Demonstrate how `continue` works in a while loop and why it is essential for controlling iteration flow.

```python
counter = 0
while counter < 10:
    counter += 1
    if counter == 4:
        continue
    print(counter)

print("\nIn a while loop, continue skips the remaining code and jumps back to the condition check.")
```

## 4. Filter Invalid Input Values

Build a small example that loops through a list of values, skips invalid entries such as `None` or negative numbers, and processes only valid values. Show a practical use case for `continue` in data validation.

```python
values = [10, 0, -2, 15, None, 7, -1, 9]
valid_values = []

for value in values:
    if value is None or value <= 0:
        continue
    valid_values.append(value)
    print(f"Processing valid value: {value}")

print(f"\nValid values: {valid_values}")
print("This is useful for filtering invalid data before processing it.")
```

## 5. Apply continue in a Nested Loop

Use nested loops to iterate over rows and columns, skipping a row or column based on a condition. Explain how `continue` affects only the innermost loop and how to structure conditions to avoid skipping too much logic.

```python
for row in range(1, 4):
    for col in range(1, 4):
        if row == 2:
            continue
        print(f"Row {row}, Col {col}")

print("\nThe continue statement skips the remaining code in the inner loop only.")
print("When row == 2, it skips all columns for that row, but does not end the entire nested loop.")
```

## 6. Compare continue and break

Create side-by-side examples that demonstrate the difference between `continue` and `break`. Show that `continue` skips the rest of the current iteration while `break` exits the loop entirely.

```python
print("continue example:")
for number in range(1, 6):
    if number == 3:
        continue
    print(number)

print("\nbreak example:")
for number in range(1, 6):
    if number == 3:
        break
    print(number)

print("\ncontinue skips the current iteration, but break exits the loop entirely.")
```

## 7. Practical Example: Process a List of Scores

Loop through a list of student scores, ignore invalid values such as -1 and values above 100, and calculate the average of the valid scores. Use `continue` to skip invalid data points and explain how this improves data-cleaning workflows.

```python
scores = [88, 92, -1, 75, 110, 67, None, 95, 81]
valid_scores = []

for score in scores:
    if score is None or score < 0 or score > 100:
        continue
    valid_scores.append(score)
    print(f"Valid score processed: {score}")

average = sum(valid_scores) / len(valid_scores)
print(f"\nAverage of valid scores: {average:.2f}")
print("This pattern is commonly used in data cleaning and validation workflows.")
```
